# Non-IID factory distributions
Run `python run.py prepare-demo` (and the matching training command) first. All numbers come from files saved by real runs.

```python
import sys, json
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
import pandas as pd
import matplotlib.pyplot as plt
from src import pipeline as pl
ctx = pl.Context(ROOT)
def load(name):
    f = ctx.results / 'metrics' / f'{name}.json'
    return json.loads(f.read_text()) if f.exists() else None
```

```python
rows = []
for i in range(ctx.cfg['federated']['num_clients']):
    info = json.loads((ctx.paths['clients']/f'client_{i+1}'/'client_info.json').read_text())
    rows.append(dict(info['class_distribution'], factory=info['name']))
df = pd.DataFrame(rows).set_index('factory')
df.plot(kind='bar', stacked=True, figsize=(7,4)); plt.ylabel('local training images'); plt.show()
df
```
